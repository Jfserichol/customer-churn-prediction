# Customer Churn Prediction.

## Model Training.

En esta fase del proyecto se entrenan distintos modelos de machine learning con el objetivo de predecir el abandono de clientes (`Churn`).

Se evaluarán diferentes algoritmos para comparar su capacidad predictiva y detectar cuál ofrece el mejor equilibrio entre rendimiento e interpretabilidad.

## Importamos librerías.

```python
import pandas as pd
import numpy as np
import os

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

pd.set_option('display.max_columns', None)
```

```python
os.makedirs(
    r"..\images",
    exist_ok=True
)
```

# Carga de datasets procesados.

```python
X_train = pd.read_csv(r"..\Data\Processed\X_train.csv")

X_test = pd.read_csv(r"..\Data\Processed\X_test.csv")

y_train = pd.read_csv(r"..\Data\Processed\y_train.csv")

y_test = pd.read_csv(r"..\Data\Processed\y_test.csv")
```

# Ajustamos el formato de la variable objetivo.

Tras cargar los datasets procesados, la variable objetivo (`y_train` y `y_test`) se convierte a formato unidimensional (`Series`) para garantizar la compatibilidad con los algoritmos de scikit-learn.

```python
y_train = y_train.squeeze()

y_test = y_test.squeeze()
```

# Verificación de dimensiones.

```python
print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

# Modelo 1: Regresión Logística.

La regresión logística se utiliza como modelo baseline debido a su interpretabilidad y amplio uso en problemas de clasificación binaria.

```python
log_model = LogisticRegression(random_state=72)

log_model.fit(X_train, y_train)
```

## Predicciones del modelo.

```python
y_pred_log = log_model.predict(X_test)

y_prob_log = log_model.predict_proba(X_test)[:,1]
```

## Evaluación de la regresión logística.

```python
print('Accuracy:', accuracy_score(y_test, y_pred_log))
print('Precision:', precision_score(y_test, y_pred_log))
print('Recall:', recall_score(y_test, y_pred_log))
print('F1 Score:', f1_score(y_test, y_pred_log))
print('ROC AUC:', roc_auc_score(y_test, y_prob_log))
```

La regresión logística obtiene un rendimiento sólido como modelo baseline, alcanzando una precisión global (`Accuracy`) superior al 80%.

El modelo presenta un equilibrio razonable entre `Precision` y `Recall`, mostrando capacidad para identificar clientes con riesgo de abandono sin generar un número excesivo de falsos positivos.

Asimismo, el valor de `ROC AUC` superior a 0.85 indica una buena capacidad de discriminación entre clientes que abandonan el servicio y aquellos que permanecen en la compañía.

En conjunto, estos resultados sugieren que la regresión logística constituye una base interpretable y competitiva para el problema de predicción de churn.

# Modelo 2: Decision Tree.

Los árboles de decisión permiten capturar relaciones no lineales entre variables y facilitan la interpretación visual de reglas de decisión.

```python
tree_model = DecisionTreeClassifier(
    random_state=72,
    max_depth=5
)

tree_model.fit(X_train, y_train)
```

## Predicciones del modelo.

```python
y_pred_tree = tree_model.predict(X_test)

y_prob_tree = tree_model.predict_proba(X_test)[:,1]
```

## Evaluación del árbol de decisión.

```python
print('Accuracy:', accuracy_score(y_test, y_pred_tree))
print('Precision:', precision_score(y_test, y_pred_tree))
print('Recall:', recall_score(y_test, y_pred_tree))
print('F1 Score:', f1_score(y_test, y_pred_tree))
print('ROC AUC:', roc_auc_score(y_test, y_prob_tree))
```

El árbol de decisión obtiene un rendimiento ligeramente inferior al observado en la regresión logística, aunque mantiene una capacidad predictiva razonable para identificar clientes con riesgo de abandono.

Se observa una disminución en métricas como `Recall` y `F1 Score`, lo que indica una menor capacidad para detectar correctamente todos los casos de churn.

No obstante, el modelo mantiene un valor de `ROC AUC` elevado, lo que sugiere que continúa diferenciando adecuadamente entre clientes que abandonan el servicio y aquellos que permanecen en la compañía.

En conjunto, el árbol de decisión permite capturar relaciones no lineales entre variables y aporta una mayor interpretabilidad estructural, aunque con un rendimiento algo más limitado.

# Modelo 3: Random Forest.

Random Forest combina múltiples árboles de decisión para mejorar la capacidad predictiva y reducir el riesgo de sobreajuste.

```python
rf_model = RandomForestClassifier(
    random_state=72,
    n_estimators=200,
    max_depth=8
)

rf_model.fit(X_train, y_train)
```

## Predicciones del modelo.

```python
y_pred_rf = rf_model.predict(X_test)

y_prob_rf = rf_model.predict_proba(X_test)[:,1]
```

## Evaluación de Random Forest.

```python
print('Accuracy:', accuracy_score(y_test, y_pred_rf))
print('Precision:', precision_score(y_test, y_pred_rf))
print('Recall:', recall_score(y_test, y_pred_rf))
print('F1 Score:', f1_score(y_test, y_pred_rf))
print('ROC AUC:', roc_auc_score(y_test, y_prob_rf))
```

Random Forest presenta un rendimiento competitivo ligeramente superior al observado en los modelos anteriores, especialmente en términos de `ROC AUC`, lo que indica una buena capacidad de discriminación entre clientes que abandonan el servicio y aquellos que permanecen en la compañía.

El modelo mejora ligeramente métricas como `Precision` y `F1 Score` respecto al árbol de decisión, reflejando una mayor robustez al combinar múltiples árboles y reducir el riesgo de sobreajuste.

No obstante, el `Recall` continúa mostrando margen de mejora, lo que sugiere que todavía existen clientes con riesgo de churn que no están siendo identificados correctamente por el modelo.

En conjunto, Random Forest ofrece un equilibrio sólido entre capacidad predictiva y estabilidad, posicionándose como uno de los modelos más prometedores hasta esta fase del proyecto.

# Modelo 4: XGBoost.

XGBoost (Extreme Gradient Boosting) es uno de los algoritmos de boosting más utilizados en problemas de clasificación tabular debido a su elevada capacidad predictiva y robustez frente al sobreajuste.

Este modelo construye múltiples árboles de forma secuencial, corrigiendo progresivamente los errores cometidos por árboles anteriores.

```python
xgb_model = XGBClassifier(
    random_state=72,
    n_estimators=200,
    learning_rate=0.05,
    max_depth=4,
    eval_metric='logloss'
)

xgb_model.fit(X_train, y_train)
```

## Predicciones del modelo.

```python
y_pred_xgb = xgb_model.predict(X_test)

y_prob_xgb = xgb_model.predict_proba(X_test)[:,1]
```

## Evaluación del XGBoost.

```python
print('Accuracy:', accuracy_score(y_test, y_pred_xgb))
print('Precision:', precision_score(y_test, y_pred_xgb))
print('Recall:', recall_score(y_test, y_pred_xgb))
print('F1 Score:', f1_score(y_test, y_pred_xgb))
print('ROC AUC:', roc_auc_score(y_test, y_prob_xgb))
```

XGBoost presenta un rendimiento sólido y competitivo, situándose entre los modelos con mejor capacidad predictiva del proyecto.

El modelo alcanza un `Accuracy` superior al 80% y mantiene un equilibrio razonable entre `Precision` y `Recall`, mostrando capacidad para identificar clientes con riesgo de abandono sin generar un número excesivo de falsos positivos.

Asimismo, el valor de `ROC AUC` cercano a 0.86 indica una elevada capacidad de discriminación entre clientes que abandonan el servicio y aquellos que permanecen activos.

En comparación con modelos anteriores, XGBoost logra mejorar métricas como `Recall` y `F1 Score` respecto al árbol de decisión y presenta un rendimiento muy similar a Random Forest, confirmando la eficacia de los métodos de boosting en problemas de clasificación tabular como la predicción de churn.

# Comparación final de modelos.

A continuación, se presentan los resultados obtenidos por los distintos modelos entrenados durante la fase de modelado.

```python
results = pd.DataFrame({
    'Model': [
        'Logistic Regression',
        'Decision Tree',
        'Random Fostest',
        'XGBoost'
    ],
    'Accuracy':[
        accuracy_score(y_test, y_pred_log),
        accuracy_score(y_test, y_pred_tree),
        accuracy_score(y_test, y_pred_rf),
        accuracy_score(y_test, y_pred_xgb)
    ],
    'Precision': [
        precision_score(y_test, y_pred_log),
        precision_score(y_test, y_pred_tree),
        precision_score(y_test, y_pred_rf),
        precision_score(y_test, y_pred_xgb)
    ],
    'Recall': [
        recall_score(y_test, y_pred_log),
        recall_score(y_test, y_pred_tree),
        recall_score(y_test, y_pred_rf),
        recall_score(y_test, y_pred_xgb)
    ],
    'F1 Score': [
        f1_score(y_test, y_pred_log),
        f1_score(y_test, y_pred_tree),
        f1_score(y_test, y_pred_rf),
        f1_score(y_test, y_pred_xgb)
    ],
    'ROC AUC': [
        roc_auc_score(y_test, y_prob_log),
        roc_auc_score(y_test, y_prob_tree),
        roc_auc_score(y_test, y_prob_rf),
        roc_auc_score(y_test, y_prob_xgb)
    ]
})

results.sort_values(by='ROC AUC', ascending=False)
```

Los modelos evaluados muestran un rendimiento relativamente similar, aunque se observan diferencias relevantes en determinadas métricas.

La regresión logística obtiene el mejor `Accuracy`, `Recall` y `F1 Score`, lo que indica una buena capacidad general para identificar clientes con riesgo de abandono manteniendo un equilibrio razonable entre falsos positivos y falsos negativos.

Por su parte, Random Forest y XGBoost presentan los valores más elevados de `ROC AUC`, reflejando una excelente capacidad de discriminación entre clientes que abandonan el servicio y aquellos que permanecen activos.

El árbol de decisión simple obtiene el rendimiento más limitado entre los modelos analizados, aunque sigue aportando interpretabilidad y capacidad para capturar relaciones no lineales.

En conjunto, los resultados sugieren que tanto la regresión logística como los modelos ensemble (`Random Forest` y `XGBoost`) constituyen alternativas sólidas para el problema de predicción de churn.

# Selección preliminar del modelo.

La selección final del modelo no debe basarse únicamente en una métrica individual, sino en el equilibrio entre capacidad predictiva, interpretabilidad y robustez.

Aunque modelos como Random Forest y XGBoost muestran una elevada capacidad de discriminación, la regresión logística mantiene un rendimiento muy competitivo y ofrece una interpretación más sencilla de los resultados.
